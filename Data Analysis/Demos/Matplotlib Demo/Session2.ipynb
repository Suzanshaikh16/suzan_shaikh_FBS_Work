```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("C:/Users/hp world karad/Downloads/Iris.csv")
data.head()
```

Scatterplot
1. Relationship between two variables
2. Correlation analysis
3. Positive correlation
4. Negative correlation

```python
data['Species'].unique()
```

```python
plt.figure(figsize=(7, 3))
plt.style.use('ggplot')
plt.scatter(data[data['Species'] == 'Setosa']['Sepal_Length'],
            data[data['Species'] == 'Setosa']['Sepal_Width'],
            color='orange', label='Setosa')
plt.scatter(data[data['Species'] == 'Versicolor']['Sepal_Length'],
            data[data['Species'] == 'Versicolor']['Sepal_Width'],
            color='red', label='Versicolor')
plt.scatter(data[data['Species'] == 'Virginica']['Sepal_Length'],
            data[data['Species'] == 'Virginica']['Sepal_Width'],
            color='blue', label='Virginica')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.legend()
plt.show()
```

#### Pie plot
1. Showing percentage distribution
2. Part-to-whole comparison

```python
subjects = ['Python','MySQL','DA','DS','WP']
marks = [90,45,50,80,70]
```

```python
plt.pie(marks, labels=subjects,
        explode=[0.05, 0.02, 0.03, 0.06, 0.04],
        autopct='%0.2f%%')
plt.show()
```
