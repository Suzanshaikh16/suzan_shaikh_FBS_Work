```python
import matplotlib.pyplot as plt
```

#### Barplot
1. comparing the categories
2. Showing frequencies
3. Product sell frequencies
4. Marks comparison
      

```python
products = ['Tea','Hot Coffee','Cold Coffee','Maggie']
summer = [60, 30, 95, 75]
```

```python
plt.bar(products,summer,color='orange')
plt.title('Sales of products in summer')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

#### Histogram
1. Distribution of numerical values
2. Frequencies of value ranges
3. Normal distribution
4. Outliers
5. Skewness

```python
import numpy as np
age = np.random.randint(1, 100, 50)
bins = [0,20, 40, 60, 80,100]
plt.hist(age, bins=bins)
plt.xlabel('Age ranges')
plt.ylabel('Frequencies')
plt.title('Age of peoples')
plt.show()
```

#### Lineplot
1. Time series analysis
2. Trends over time
3. Upward/Downward trends
4. IPL team scoresheet

```python
overs= np.arange(1, 21)
team1_runs = np.random.randint(0, 37, 20)
team2_runs = np.random.randint(0, 37, 20)
```

```python
team1_score = np.cumsum(team1_runs)
team2_score = np.cumsum(team2_runs)
```

```python
team1_runs
```

```python
team1_score
```

```python
plt.style.use('dark_background')
plt.plot(overs, team1_score, color='red', label='RCB') 
plt.plot(overs, team2_score, color='blue', label='MI')
plt.xlabel('Overs') 
plt.ylabel('Runs')
plt.title('Scoresheet')
plt.legend()
plt.show()
```
