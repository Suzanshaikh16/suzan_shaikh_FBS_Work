```python
emp = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Name': ['Suzan', 'Suhana', 'Saima', 'Aalisha'],
    'Age': [34, 56, 55, 29],
    'DID':[1,3,1,2]
})
emp
```

```python
dept = pd.DataFrame({
    'DID': [1, 2, 3, 4],
    'DName': ['IT', 'Admin', 'Hr', 'Sales']})
dept
```

```python
pd.merge(emp,dept)
```

```python
dept2 = pd.DataFrame({'ID':[1,2,3,4],
                     'DName':['IT','Admin','Hr','Sales']})
dept2
```

```python
pd.merge(emp,dept2,left_on='DID',right_on='ID')
```

```python
pd.merge(emp,dept,how='left')
```

```python
pd.merge(emp,dept,how='right')
```

```python
pd.merge(emp,dept,how='outer')
```
